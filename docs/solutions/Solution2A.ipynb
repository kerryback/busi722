```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_raw = pd.read_parquet("merged.parquet")
print(f"Loaded: {len(df_raw):,} rows, {df_raw.shape[1]} columns")
print(f"Date range: {df_raw['month'].min()} to {df_raw['month'].max()}")
```

## 1. Report totals

```python
print(f"Total rows: {len(df_raw):,}")
print(f"Unique tickers: {df_raw['ticker'].nunique()}")
print(f"Date range: {df_raw['month'].min()} to {df_raw['month'].max()}")
```

## 2. Penny-stock filter

```python
before = len(df_raw)
df_penny = df_raw[df_raw["close"] >= 5.0].copy()
after = len(df_penny)
print(f"Rows before penny filter: {before:,}")
print(f"Rows after close >= $5:   {after:,}")
print(f"Rows dropped:             {before - after:,} ({(before-after)/before:.1%})")
```

## 3. Missing momentum

```python
mom_missing = df_raw["momentum"].isna().sum()
print(f"Missing momentum: {mom_missing:,} / {len(df_raw):,} ({mom_missing/len(df_raw):.1%})")
```

Momentum is missing for early observations because it requires 13 months of prior price history (cumulative return from $t-13$ to $t-2$). The first 13 months of each stock's history will have missing momentum.
