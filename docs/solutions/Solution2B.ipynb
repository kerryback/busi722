```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_raw = pd.read_parquet("merged.parquet")
print(f"Loaded: {len(df_raw):,} rows, {df_raw.shape[1]} columns")
print(f"Date range: {df_raw['month'].min()} to {df_raw['month'].max()}")
```

## 1. Asset growth ordering

`asset_growth` must be computed **before** merging with monthly returns. After forward-filling, consecutive months share the same fundamental values. If you compute `pct_change()` after forward-fill, growth would be 0 for all filled months. Computing on raw SF1 data (annual filings) gives the true year-over-year growth rate.

## 2-3. Fundamental variables

```python
fund_vars = ["roe", "grossmargin", "assetturnover", "de", "asset_growth", "gp_to_assets"]
print("Fundamental variable coverage:")
for v in fund_vars:
    if v in df_raw.columns:
        n = df_raw[v].notna().sum()
        print(f"  {v:20s}: {n:>10,} non-missing ({n/len(df_raw):.1%})")

print(f"\nUnique tickers: {df_raw['ticker'].nunique()}")
print(f"Date range: {df_raw['month'].min()} to {df_raw['month'].max()}")
```
